# Expense Prediction - Data Analysis

This notebook analyzes the transaction dataset used for training the Expense Prediction Machine Learning model.

## Import libraries

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

## Load Dataset

```python
df = pd.read_csv("../datasets/transactions.csv")
df.head()
```

## Dataset Information

```python
df.info()
```

## Missing Values

```python
df.isnull().sum()
```

## Statistical Summary

```python
df.describe()
```

## Categorical Distribution

```python
df["category"].value_counts()
```

## Payment Method Distribution

```python
df["payment_method"].value_counts()
```

## Average Expense by Category

```python
df.groupby("category")["amount"].mean()
```

## Income Distribution

```python
plt.hist(df["monthly_income"], bins=20)

plt.title("Monthly Income Distribution")
plt.xlabel("Monthly Income")
plt.ylabel("Frequency")

plt.show()
```

## Expense Distribution

```python
plt.hist(df["amount"], bins=20)

plt.title("Expense Distribution")
plt.xlabel("Expense Amount")
plt.ylabel("Frequency")

plt.show()
```
